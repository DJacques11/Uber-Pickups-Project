```python
import pandas as pd

import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

# pio.templates.default = "?"
# pio.renderers.default = "vscode"
```

```python
dataset = pd.read_csv('uber-trip-data/uber-raw-data-aug14.csv')

dataset = dataset.drop('Base', axis=1)
dataset.head()
```

```python
dataset['Date/Time'] = pd.to_datetime(dataset['Date/Time'])
dataset.shape
```

```python
mask = dataset['Date/Time'].between('8/1/2014 16:00:00', '8/1/2014 17:00:00')
dataset = dataset[mask]
print(dataset.shape)
print()
print(dataset.head())
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X = sc.fit_transform(dataset[['Lat', 'Lon']].values)

X[:5]
```

```python
#ELBOW METHOD
from sklearn.cluster import KMeans

wcss = []
k = []

for i in range(1, 20):
    kmeans = KMeans(n_clusters=i, random_state=0, n_init='auto')
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
    k.append(i)

    print('WCSS pour K={} --> {}'.format(i, wcss[-1]))
```
